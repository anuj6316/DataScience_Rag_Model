````python
import json
from pathlib import Path
from typing import List, Dict

# Docling imports
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions

class PdfRagProcessor:
    def __init__(self, output_dir: str = "./rag_output"):
        self.output_dir = Path(output_dir)
        self.output_dir.mkdir(parents=True, exist_ok=True)
        
        # --- CONFIGURATION ---
        pipeline_options = PdfPipelineOptions(
            do_ocr=True,
            do_table_structure=True,
            generate_page_images=False,    # Context backup (full page)
            generate_picture_images=True, # The specific diagrams/architectures
            images_scale=2.0              # High res for readability
        )

        self.converter = DocumentConverter(
            format_options={
                InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
            }
        )

    def process_pdf(self, pdf_path: str) -> List[Dict]:
        file_path = Path(pdf_path)
        print(f"Processing: {file_path.name}...")
        
        # 1. Convert
        result = self.converter.convert(file_path)
        doc = result.document
        
        rag_chunks = []

        # 2. Iterate through pages
        for page_no, page in doc.pages.items():
            
            # --- A. Extract Text & Code ---
            page_markdown = doc.export_to_markdown(page_no=page_no)
            
            # --- B. Extract Architecture/Flowcharts (Images) ---
            extracted_diagram_paths = []
            
            for i, picture in enumerate(doc.pictures):
                # Check if this picture belongs to the current page
                if picture.prov and picture.prov[0].page_no == page_no:
                    image_filename = f"{file_path.stem}_p{page_no}_diag{i}.png"
                    image_path = self.output_dir / image_filename
                    
                    # --- FIX: Access .pil_image inside the ImageRef wrapper ---
                    if picture.image and picture.image.pil_image:
                        with open(image_path, "wb") as fp:
                            picture.image.pil_image.save(fp, format="PNG")
                        extracted_diagram_paths.append(str(image_path))

            # --- C. Extract Full Page Reference (Backup) ---
            full_page_image_path = self.output_dir / f"{file_path.stem}_full_p{page_no}.png"
            
            # --- FIX: Access .pil_image inside the ImageRef wrapper ---
            if page.image and page.image.pil_image:
                with open(full_page_image_path, "wb") as fp:
                    page.image.pil_image.save(fp, format="PNG")

            # --- D. Build RAG Record ---
            chunk = {
                "source_file": file_path.name,
                "page_number": page_no,
                "text_content": page_markdown,
                "diagram_images": extracted_diagram_paths,
                "full_page_image": str(full_page_image_path),
                "has_code": "```" in page_markdown
            }
            
            rag_chunks.append(chunk)

        return rag_chunks

# --- EXECUTION ---
if __name__ == "__main__":
    # Update this path to your actual PDF location
    pdf_file = "/home/anuj/DataScience_Rag_Model/PDF document for PF Chatbot.pdf" 
    
    processor = PdfRagProcessor()
    
    try:
        chunks = processor.process_pdf(pdf_file)
        
        # Save metadata to JSON
        with open(processor.output_dir / "rag_db_input.json", "w") as f:
            json.dump(chunks, f, indent=2)
        
        print(f"Success! Processed {len(chunks)} pages.")
        print(f"Data saved to: {processor.output_dir.absolute()}")
        
    except Exception as e:
        print(f"Error: {e}")
        import traceback
        traceback.print_exc()
````
